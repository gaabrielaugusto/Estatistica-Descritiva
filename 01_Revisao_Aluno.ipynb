### Execute o Script para Gerar o Arquivo CSV de Teste (~1.000 linhas)
Executa este script para construir a base sintética que simula o banco de dados da empresa.

```python
import pandas as pd # biblioteca para manipulação de dados 
import numpy as np # biblioteca para operações numéricas

# Definindo número aleatório para reprodutibilidade dos dados
np.random.seed(42)

# Definindo o volume de linhas do arquivo
TOTAL_LINHAS = 1000

# Lista de categorias e produtos para sorteio
categorias = ['Eletrônicos', 'Móveis', 'Informática', 'Eletrodomésticos']
cidades = ['Rio de Janeiro', 'Niterói', 'Petrópolis', 'Volta Redonda', 'Macaé']

# Gerando dados sintéticos
dados_sinteticos = {
    'ID_Transacao': [f"TXN-{10000 + i}" for i in range(TOTAL_LINHAS)],
    'Data_Venda': pd.date_range(start='2024-01-01', periods=TOTAL_LINHAS, freq='4h'),
    'Cidade': np.random.choice(cidades, size=TOTAL_LINHAS, p=[0.4, 0.2, 0.15, 0.15, 0.1]),
    'Categoria': np.random.choice(categorias, size=TOTAL_LINHAS),
    'Quantidade': np.random.randint(1, 10, size=TOTAL_LINHAS),
    'Preco_Unitario': np.round(np.random.uniform(50.0, 3500.0, size=TOTAL_LINHAS), 2)
}

# Criando o DataFrame
df_gerado = pd.DataFrame(dados_sinteticos)

# Calculando a coluna do Valor Total
df_gerado['Valor_Total'] = df_gerado['Quantidade'] * df_gerado['Preco_Unitario']

```

Exercício 2: Análise de Notas de Alunos
Enunciado:
Desenvolva um programa que analise as notas de alunos. O programa deve ser estruturado em funções e seguir as instruções abaixo:

Função le_dados_alunos(nome_arquivo): Esta função deve ler um arquivo CSV contendo informações de alunos com as colunas: Nome, Nota1, Nota2, Nota3 e retornar um DataFrame.

Função calcula_media(df): Esta função deve calcular a média das notas de cada aluno e adicionar uma nova coluna chamada Media ao DataFrame.

Função verifica_aprovacao(df): Esta função deve adicionar uma nova coluna chamada Aprovado, que deve conter True se a média do aluno for maior ou igual a 7 e False caso contrário.

Função separa_alunos_aprovados_reprovados(df): Esta função deve separar os alunos em dois DataFrames: um para aprovados e outro para reprovados, com base na coluna Aprovado.

Função analise_notas(nome_arquivo): Esta função deve orquestrar as chamadas das funções anteriores e retornar os DataFrames de alunos aprovados e reprovados.

```python
import pandas as pd


caminho_alunos = 'c:\\Users\\gabri\\OneDrive\\Documentos\\Gabriel\\Aula Python\\UC2\\Data\\notas.csv'
def le_dados_alunos(caminho_alunos):
    df_aluno = pd.read_csv(caminho_alunos)
    return df_aluno

dfcurso = le_dados_alunos(caminho_alunos)
print(dfcurso)
print('='*120)

print('\n--- Média das Notas ---')
def calcula_media(dfcurso):
    dfcurso['Media'] = (dfcurso['Nota1'] + dfcurso['Nota2'] + dfcurso['Nota3'])/3
    dfcurso['Media'] = dfcurso['Media'].round(2)
    return dfcurso

df_media = calcula_media(dfcurso)
print(df_media)
print('='*120)

print('\n --- Verificação de Aprovação')
def verifica_aprovacao(dfcurso):
    dfcurso['Aprovado'] = dfcurso['Media'] >= 7
    return dfcurso

df_aprovacao = verifica_aprovacao(dfcurso)
print(df_aprovacao)
print('='*120)

print('\n--- Alunos Aprovados e Reprovados ---')
def separa_aprovados_reprovados(dfcurso):
    df_aprovados = dfcurso[dfcurso['Aprovado'] == True]
    df_reprovados = dfcurso[dfcurso['Aprovado'] == False]
    return df_aprovados, df_reprovados

df_aprovados, df_reprovados = separa_aprovados_reprovados(df_aprovacao)
print('\n--- Alunos Aprovados ---')
print(df_aprovados)

print('\n--- Alunos Reprovados ---')
print(df_reprovados)
print('='*120)

def analise_notas(caminho_alunos):
    dfcurso = le_dados_alunos(caminho_alunos)
    df_media = calcula_media(dfcurso)
    df_aprovacao = verifica_aprovacao(dfcurso)
    df_aprovados, df_reprovados = separa_aprovados_reprovados(df_aprovacao)
    return df_aprovados, df_reprovados

    
```

Exercício 3: Análise de Dados de Funcionários
Enunciado:
Crie um programa que analise os dados de funcionários de uma empresa. O programa deve ser estruturado em funções e seguir as instruções abaixo:

Função le_dados_funcionarios(nome_arquivo): Esta função deve ler um arquivo CSV com as colunas: Nome, Departamento, Salário e retornar um DataFrame.

Função calcula_salario_medio(df): Esta função deve agrupar os dados por departamento e calcular o salário médio de cada um, retornando um novo DataFrame com o nome do departamento e o salário médio.

Função filtra_departamentos_alto_salario(df, limite): Esta função deve filtrar o DataFrame para retornar apenas os departamentos cuja média salarial é superior a um valor limite passado como argumento.

Função analise_funcionarios(nome_arquivo): Esta função deve orquestrar as chamadas das funções anteriores e retornar o DataFrame final com os departamentos que têm um salário médio acima de R$ 5000,00.

```python
import pandas as pd

caminho_funcionarios = 'c:\\Users\\gabri\\OneDrive\\Documentos\\Gabriel\\Aula Python\\UC2\\Data\\funcionarios.csv'
def le_dados_funcionarios(caminho_funcionarios):
    df_func = pd.read_csv(caminho_funcionarios, sep=',', encoding='utf-8')
    return df_func

dftrabalho = le_dados_funcionarios(caminho_funcionarios)
print(dftrabalho)
print('='*120)

def calcula_salario_medio(dftrabalho):
    df_medio = dftrabalho.groupby('Departamento')['Salário'].mean().reset_index()
    df_medio['Salário'] = df_medio['Salário'].round(2)
    return df_medio
df_resultado = calcula_salario_medio(dftrabalho)
print('\n--- Salário Médio por Departamento ---')
print(df_resultado)
print('='*120)

def filtra_departamento_alto_salario(dftrabalho,limite):
    df_alto_salario = dftrabalho[dftrabalho['Salário']> limite]
    return df_alto_salario

df_departamento_rico = filtra_departamento_alto_salario(df_resultado,5000)
print('\n--- Departamento com Média Acima de 5.000 ---')


def analise_funcionarios(caminho_funcionarios):
    dftrabalho = le_dados_funcionarios(caminho_funcionarios)
    df_resultado = calcula_salario_medio(dftrabalho)
    df_departamento_rico = filtra_departamento_alto_salario(df_resultado,5000)

    return df_departamento_rico

df_final = analise_funcionarios(caminho_funcionarios)
print(df_final)


```

O programa tem três funções auxiliares, e a função main() chama as três: uma para ler o arquivo, outra para calcular as receitas e uma terceira para resumir e filtrar os resultados.

```python
import pandas as pd

def ler_arquivo(caminho_do_arquivo):
    return pd.read_csv(caminho_do_arquivo, sep=',', encoding='utf-8')

def calcular_receitas(dfcompleto):
    dfcompleto['Receita Total'] = dfcompleto['Quantidade'] * dfcompleto['Preco']
    return dfcompleto

def filtrar_resumir(dfcompleto, limite):
    df_filtrado = dfcompleto[dfcompleto['Receita Total'] > limite]
    df_resumido = df_filtrado.groupby('Produto')['Receita Total'].sum().reset_index()
    df_resumido['Receita Total'] = df_resumido['Receita Total'].round(2)
    return df_resumido

def main():
    caminho_do_arquivo = 'c:\\Users\\gabri\\OneDrive\\Documentos\\Gabriel\\Aula Python\\UC2\\Data\\vendas.csv'
    print(f'Lendo o arquivo: {caminho_do_arquivo}')
    df_vendas = ler_arquivo(caminho_do_arquivo)
    print('Calculando Receitas')
    df_calculados = calcular_receitas(df_vendas)
    print('Filtrando e Resumindo')
    df_final = filtrar_resumir(df_calculados, 100)
    print('Resultado Final')
    print(df_final)

if __name__== '__main__':
    main()
```
